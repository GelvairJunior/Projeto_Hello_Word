```python
def cima(u):
    left(90)
    forward(u)
    right(90)

def baixo(u):
    right(90)
    forward(u)
    left(90)

def esquerda(u):
    left(180)
    forward(u)
    right(180)

def direita(u):
    forward(u)

def canto():
    penup()
    forward(150)
    left(90)
    forward(150/2)
    pendown()
    right(90)

```

```python
SquareSize = 512
make_turtle(delay=0.001)

u = 5

def A(i):
    if i > 0:
        D(i - 1)
        
        esquerda(u)

        A(i - 1)
        
        baixo(u)

        A(i - 1)

        direita(u)

        B(i - 1)

def B(i):
    if i > 0:
        
        C(i - 1)

        cima(u)

        B(i - 1)

        direita(u)

        B(i - 1)

        baixo(u)

        A(i - 1)

def C(i):
    if i > 0:
        B(i - 1)

        direita(u)

        C(i - 1)

        cima(u)

        C(i - 1)

        esquerda(u)

        D(i - 1)

def D(i):
    if i > 0:
        A(i - 1)
        
        baixo(u)

        D(i - 1)
        
        esquerda(u)

        D(i - 1)

        cima(u)

        C(i - 1)

canto()
A(0)
```
