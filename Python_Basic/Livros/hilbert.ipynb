```python
from multiprocessing.reduction import steal_handle

from jupyturtle import *
import math
```

```python
def cima(u):
    left(90)
    forward(u)
    right(90)

def baixo(u):
    right(90)
    forward(u)
    left(90)

def esquerda(u):
    left(180)
    forward(u)
    right(180)

def direita(u):
    forward(u)

def canto():
    penup()
    forward(150)
    left(90)
    forward(150/2)
    pendown()
    right(90)

```

```python
SquareSize = 512
make_turtle(delay=0.001)

u = 5

def A(i):
    if i > 0:
        D(i - 1)
        
        esquerda(u)

        A(i - 1)
        
        baixo(u)

        A(i - 1)

        direita(u)

        B(i - 1)

def B(i):
    if i > 0:
        
        C(i - 1)

        cima(u)

        B(i - 1)

        direita(u)

        B(i - 1)

        baixo(u)

        A(i - 1)

def C(i):
    if i > 0:
        B(i - 1)

        direita(u)

        C(i - 1)

        cima(u)

        C(i - 1)

        esquerda(u)

        D(i - 1)

def D(i):
    if i > 0:
        A(i - 1)
        
        baixo(u)

        D(i - 1)
        
        esquerda(u)

        D(i - 1)

        cima(u)

        C(i - 1)

canto()
A(5)
```

```python
def desenha(i, u):
    left(i*45)
    forward(u)
    right(i*45)

def cantoDiretio():
    penup()
    left(180)
    forward(150)
    right(90)
    forward(150/2)
    pendown()
    right(90)
```

```python
make_turtle(delay=0.001)
u = 5
def As(k):
    if k > 0:
        As(k-1)
        desenha(7, u)
        Bs(k-1)
        desenha(0, u*2)
        Ds(k-1)
        desenha(1, u)
        As(k-1)

def Bs(k):
    if k > 0:
        Bs(k-1)
        desenha(5, u)
        Cs(k-1)
        desenha(6, 2*u)
        As(k-1)
        desenha(7, u)
        Bs(k-1)

def Cs(k):
    if k > 0:
        Cs(k-1)
        desenha(3, u)
        Ds(k-1)
        desenha(4, 2*u)
        Bs(k-1)
        desenha(5, u)
        Cs(k-1)

def Ds(k):
    if k > 0:
        Ds(k-1)
        desenha(1, u)
        As(k-1)
        desenha(2, 2*u)
        Cs(k-1)
        desenha(3, u)
        Ds(k-1)

def sierpinski(s):
    As(s)
    desenha(7, u)
    Bs(s)
    desenha(5, u)
    Cs(s)
    desenha(3, u)
    Ds(s)
    desenha(1, u)
cantoDiretio()
sierpinski(3)

```
