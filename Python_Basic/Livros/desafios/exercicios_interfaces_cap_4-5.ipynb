```python
import turtle
from multiprocessing.reduction import steal_handle
from jupyturtle import *
import math


def parellelogram(altura, largura, angule, angulo_oposto):

    for i in range(2):
        forward(largura)
        left(angule)
        forward(altura)
        left(angulo_oposto)

def rastro():
    penup()
    left(180)
    forward(50)
    left(90)
    forward(30)
    pendown()
    left(90)

def retangle(altura, largura):
    parellelogram(altura, largura, 90, 90)

def losangulo(comprimento, angule):
    angulo_oposto = 180 - angule
    parellelogram(comprimento, comprimento, angule, angulo_oposto)

make_turtle()

rastro()

retangle(50, 80)
```

```python

def triangule(base, lado):
    # Calcula o ângulo da base
    angulo_base = math.degrees(math.acos(base / (2 * lado)))

    # Primeiro lado
    right(180 - angulo_base)
    forward(lado)

    # Segundo lado
    right(2 * angulo_base)
    forward(lado)

    # Desenha a base
    right(180 - angulo_base)
    forward(base)

def poligonos_triangulos(lados, tamanho):
    raio = tamanho/ (2 * math.sin(math.radians(180 / lados)))
    angulo = 360/lados

    for i in range(lados):
        forward(tamanho)
        triangule(tamanho, raio)
        right(angulo)

make_turtle(delay=0.001)
poligonos_triangulos(100, 10)
```

```python
def arc(radius, angle):
    arc_length = 2 * math.pi * radius * angle /360
    n=30
    length = arc_length / n
    poligono(n, length)

def poligono(lados, R):
    angulo = 360/lados
    for i in range(lados):
        forward(R)
        right(angulo)

def espiral(tamanho):
    length = 1

    for i in range(tamanho):
        forward(length)
        left(10)
        length += 0.2  # aumenta o passo

make_turtle(delay=0.001)
rastro()
espiral(100)
```

```python
def arc(radius, angle):
    arc_length = 2 * math.pi * radius * angle /360
    n=30
    length = arc_length / n
    step_angle = angle/n
    polyLinhas(n, length, step_angle)

def polyLinhas(n, length, angle):
    for i in range(n):
        forward(length)
        left(angle)

def petal():
    arc(40, 180)

make_turtle(delay=0.001)
rastro()
petal()
```

```python
def draw(length):
    angle = 50
    factor = 0.6

    if length > 5:
        forward(length)
        left(angle)
        draw(factor * length)
        right(2 * angle)
        draw(factor * length)
        left(angle)
        back(length)

make_turtle()
draw(50) # arvore fractural, ela divide-se em diversos galhos binarios tendo 60% do seu tamanho original, os galhos sempre se dividem em dois e criam novos galhos até atingir o valor minimo que é 5 e volta para sua posição original
```

```python
def koch(length, nivel):
    esq = 60
    dir = 120

    if nivel == 0:
        forward(length)
    else:
        koch(length/3, nivel-1)
        left(esq)
        koch(length/3, nivel-1)
        right(dir)
        koch(length/3, nivel-1)
        left(esq)
        koch(length/3, nivel-1)

make_turtle(delay=0.001)
koch(120, 4)
```

```python
def triangulo_equilatero(tamanho):
    for i in range(3):
        forward(tamanho)
        left(120)

def draw_pie(lados, R):
    angulo = 360/lados
    for i in range(lados):
        forward(R)
        triangulo_equilatero(R)
        right(angulo)

def sierpinski(tamanho, nivel):
    if nivel == 0:
        triangulo_equilatero(tamanho)
    else:
        sierpinski(tamanho / 2, nivel - 1)

        forward(tamanho / 2)
        sierpinski(tamanho / 2, nivel - 1)

        back(tamanho / 2)
        left(60)

        forward(tamanho / 2)
        right(60)

        sierpinski(tamanho / 2, nivel - 1)

        left(60)
        back(tamanho / 2)
        right(60)

make_turtle(delay=0.000)
rastro()
sierpinski(100, 3)
```
