```python
from copy import copy, deepcopy
from jupyturtle import make_turtle, jumpto, moveto
```

```python
class Point:

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f'Point({self.x}, {self.y})'

    def __eq__(self, other):
        return (self.x == other.x) and (self.y == other.y)

    def translate(self, dx=0, dy=0):
        self.x += dx
        self.y += dy

    def translated(self, dx=0, dy=0):
        point = copy(self)
        point.translate(dx, dy)
        return point

class Line:
    def __init__(self, p1, p2):
        self.p1 = p1
        self.p2 = p2

    def __str__(self):
        return f'Line({self.p1}, {self.p2})'

    def __eq__(self, other): # desafio 1
        return self.p1 is other.p1 or self.p2 is other.p2

    def draw(self):
        jumpto(self.p1.x, self.p1.y)
        moveto(self.p2.x, self.p2.y)

    def midpoint(self): # desafio 2
        return Point((self.p1.x + self.p2.x)/2, (self.p1.y + self.p2.y) /2)

class Rectangle:
    def __init__(self, width, height, corner):
        self.width = width
        self.height = height
        self.corner = corner

    def __str__(self):
        return f'Rectangle({self.width}, {self.height}, {self.corner})'

    def make_points(self):
        p1 = self.corner
        p2 = p1.translated(self.width, 0)
        p3 = p2.translated(0, self.height)
        p4 = p3.translated(-self.width, 0)
        return p1, p2, p3, p4

    def make_lines(self):
        p1,p2,p3,p4 = self.make_points()
        return Line(p1, p2), Line(p2, p3), Line(p3, p4), Line(p4, p1)

    def draw(self):
        lines = self.make_lines()
        for line in lines:
            line.draw()

    def grow(self, dwidth, dheight):
        self.width += dwidth
        self.height += dheight

    def translate(self, dx, dy):
        self.corner.translate(dx, dy)

    def midpoint(self): # desafio 3
        l1, l2, l3, l4 = self.make_lines()
        return Point(l1.midpoint().x, l2.midpoint().y)

    def make_cross(self): # desafio 4
        l1, l2, l3, l4 = self.make_lines()
        p1 = l1.midpoint()
        p2 = l3.midpoint()
        p3 = l2.midpoint()
        p4 = l4.midpoint()
        return [Line(p1, p2), Line(p3, p4)]
        ...

class Circle: # desafio 5

    def __init__(self, center, radius):
        self.center = center
        self.radius = radius

    def __str__(self):
        return f'Circulo({self.center}, raio ({self.radius}) )'

    def draw(self):
        ...
```
