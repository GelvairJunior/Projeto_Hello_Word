```python
from multiprocessing.reduction import steal_handle

from jupyturtle import left, right, forward, make_turtle
import math
```

```python

def polyLinhas(n, length, angle):
    for i in range(n):
        forward(length)
        left(angle)

def poligono(n, length):
    angle = 360 / n
    polyLinhas(n, length, angle)

def arc(radius, angle):
    arc_length = 2 * math.pi * radius * angle /360
    n=30
    length = arc_length / n
    stepangle = angle / n
    polyLinhas(n, length, stepangle)

def circle(radius):
    arc(radius, 360)

```

```python
make_turtle()
poligono(n=7, length=40)
left(45)
poligono(6, 20)
```

```python
make_turtle(delay=0.001)
arc(50, 360)
```

```python
make_turtle(delay=0)
poligono(20, 9)
arc(70, 70)
circle(10)
```
